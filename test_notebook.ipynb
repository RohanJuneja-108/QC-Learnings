```python
import qiskit
from qiskit import QuantumCircuit

# Create a simple quantum circuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)

print(f"Qiskit version: {qiskit.__version__}")
print("Jupyter Notebook is working perfectly!")
```

```python
# Draw the circuit visually using matplotlib
qc.draw('mpl') 
```
